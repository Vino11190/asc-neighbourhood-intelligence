```python
import pandas as pd 
import matplotlib.pyplot as plt 

months = pd.date_range('2024-01-01', periods=24, freq='ME') 

demand = pd.DataFrame({ 
    'Month': months, 
    'Referrals': [ 980, 995, 1005, 1010, 1025, 1040, 1055, 1060, 1075, 1090, 1105, 1120, 1135, 1145, 1160, 1175, 1190, 1205, 1220, 1235, 1250, 1265, 1280, 1295 ] }) 
    
print(demand.head())
```

```python
plt.figure(figsize=(10,5)) 
plt.plot(demand['Month'], demand['Referrals'], marker='o') 
plt.title('Monthly Adult Social Care Referrals') 
plt.xlabel('Month') 
plt.ylabel('Referrals') 
plt.grid(True) 
plt.tight_layout() 

plt.savefig('../outputs/monthly_demand_trend.png') 

plt.show()
```

```python
last_value = demand['Referrals'].iloc[-1] 
future_months = pd.date_range( 
    demand['Month'].iloc[-1] + pd.offsets.MonthEnd(1), 
    periods=12, freq='ME' ) 

forecast_values = [last_value + (i+1)*15 for i in range(12)] 

forecast = pd.DataFrame({ 
    'Month': future_months, 'Forecast_Referrals': forecast_values }) 
    
print(forecast.head())
```

```python
plt.figure(figsize=(11,5)) 
plt.plot(demand['Month'], demand['Referrals'], label='Actual', marker='o') 
plt.plot(forecast['Month'], forecast['Forecast_Referrals'], label='Forecast', linestyle='--', marker='o') 
plt.title('Adult Social Care Demand Forecast') 
plt.xlabel('Month') 
plt.ylabel('Referrals') 
plt.legend() 
plt.grid(True) 
plt.tight_layout() 

plt.savefig('../outputs/demand_forecast.png') 
plt.show()
```

```python
current = demand['Referrals'].iloc[-1] 
future = forecast['Forecast_Referrals'].iloc[-1] 

increase_pct = round((future - current) / current * 100, 1) 

print(f'Projected increase over 12 months: {increase_pct}%')
```

Forecast insight

The demand forecast indicates a projected 13.9% increase in Adult Social Care referrals over the next 12 months. If this trend continues, additional capacity may be required across assessment, reablement and community support services. The forecast supports proactive planning and provides an evidence base for transformation and neighbourhood-based prevention initiatives.
