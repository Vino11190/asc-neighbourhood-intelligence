```python
import pandas as pd 
asc = pd.DataFrame({ 
    'LSOA_CODE': [ 'E01032739','E01032740','E01032741','E01032742','E01032743', 
                  'E01032744','E01032745','E01032746','E01032747','E01032748' 
                  ], 

    'LSOA_NAME': [ 'Southampton 001A','Southampton 001B','Southampton 002A','Southampton 002B', 
                  'Southampton 003A','Southampton 003B','Southampton 004A','Southampton 004B', 
                  'Southampton 005A','Southampton 005B' ], 

    'ASC_Referrals': [120,95,140,110,160,130,175,145,155,125], 
    'Reablement_Cases': [45,38,52,41,60,48,66,54,58,46] }) 

print(asc.head()) 
print('ASC columns:', len(asc.columns)) 
print('ASC rows:', len(asc))
```

```python
pop = pd.DataFrame({ 
    'LSOA_CODE': [ 'E01032739','E01032740','E01032741','E01032742','E01032743', 
                  'E01032744','E01032745','E01032746','E01032747','E01032748' 
                  ], 
    'LSOA_NAME': [ 'Southampton 001A','Southampton 001B','Southampton 002A','Southampton 002B', 
                  'Southampton 003A','Southampton 003B','Southampton 004A','Southampton 004B', 
                  'Southampton 005A','Southampton 005B' 
                  ], 
    'Population_65_plus': [420,365,510,445,590,470,640,530,575,455] }) 

print(pop.head()) 
print('Population columns:', len(pop.columns)) 
print('Population rows:', len(pop))
```

```python
merged = asc.merge(pop, on=['LSOA_CODE', 'LSOA_NAME']) 

print(merged.head()) 
print('Merged columns:', len(merged.columns)) 
print('Merged rows:', len(merged))
```

```python
merged['Referral_Rate_per_1000'] = ( 
    merged['ASC_Referrals'] / merged['Population_65_plus'] * 1000 
    ).round(1) 

result = merged[['LSOA_NAME', 'ASC_Referrals', 'Population_65_plus', 'Referral_Rate_per_1000']] 

print(result)
```

```python
merged.to_csv('../outputs/asc_integrated_dataset.csv', index=False) 

print('Integrated dataset saved successfully')
```
